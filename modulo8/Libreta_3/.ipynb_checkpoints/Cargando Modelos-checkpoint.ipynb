```python
import tensorflow as tf
import numpy as np
```

```python
#Cargamos la red neuronal
model = tf.keras.models.load_model('my_model.h5')
#Cargamos los datos de prueba
x_test = np.load("xtest.npy")
y_test = np.load("ytest.npy")
```

```python
#Evaluamos con los mismos datos

test_loss = model.evaluate(x_test, y_test)
print('Test loss:', test_loss)
```

## Cargando redes pre-entrenadas

Algunas redes pre-entrendas: https://keras.io/applications/

```python
from tensorflow.keras.applications.resnet50 import ResNet50
from tensorflow.keras.preprocessing import image
from tensorflow.keras.applications.resnet50 import preprocess_input, decode_predictions

model = ResNet50()
```

```python
# https://unsplash.com/photos/Bj0XFTabquo
img_path = 'data/elephant.jpg'

img = image.load_img(img_path, target_size=(224, 224)) 
x = image.img_to_array(img)
x = np.expand_dims(x, axis=0)
x = preprocess_input(x)
```

```python
preds = model.predict(x)
print('Predicciones:', )
for n in  decode_predictions(preds)[0]:
    print(n[1:])
```
